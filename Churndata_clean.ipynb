```python
df.drop_duplicates(inplace=True)
```

```python
df.replace(['N/A','NULL',' ',''],np.nan,inplace=True)
```

```python
df.apply(lambda x: x.str.strip() if x.dtype=='str' else x)
```

```python
df['Customer_Name']=df['Customer_Name'].str.title()
df['State']=df['State'].str.title()
df['City']=df['City'].str.title()
df['Subscription_Type']=df['Subscription_Type'].str.title()
df['Contract_Type']=df['Contract_Type'].str.title()
```

```python
df['Subscription_Type'].value_counts()
```

```python
df['Subscription_Type']=df['Subscription_Type'].str.strip()
```

```python
df['Subscription_Type'].value_counts()
```

```python
df['Churn'].value_counts()
```

```python
df['Churn']=df['Churn'].str.title()
df['Gender']=df['Gender'].str.title()
```

```python
df['Gender'].value_counts()
```

```python
df['Gender']=df['Gender'].str.strip()
```

```python
df['Gender'].value_counts()
```

```python
df=df[(df['Age']>=18) & (df['Age']<=100)]
```

```python
df['Age'].max()
```

```python
df=df[df['Monthly_Charges']>=0]
df=df[df['Total_Charges']>=0]
```

```python
df.shape
```

```python
df.columns
```

```python
cols=['Age','Tenure_Months','Monthly_Charges','Total_Charges']

for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
df['Last_Interaction_Date'].head(200)
```

```python
df['Last_Interaction_Date']=pd.to_datetime(df['Last_Interaction_Date'],format='%d-%m-%y',errors='coerce'op
```

```python
df['Last_Interaction_Date'].isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Tech_Support'].fillna('Unknown',inplace=True)
df['Payment_Method'].fillna('Unknown',inplace=True)
df['Monthly_Charges'].fillna(df['Monthly_Charges'].mean(),inplace=True)
    df['Tenure_Months'].fillna(df['Tenure_Months'].mean(),inplace=True)
```

```python
df.shape
```

```python
df['Customer_value']=df['Monthly_Charges']*df['Tenure_Months']
```

```python
df.head()
```

```python
df['Monthly_revenue']=df['Monthly_Charges']
```

```python
df.shape
```

```python
df.drop('Monthly_Charges', axis=1, inplace=True)
```

```python
df
```

```python
bins=[0,12,24,48,72]

label=labels=['0-12 month','12-24 month','24-48 month','48-72 month']
df['Tenure_group']=pd.cut(df['Tenure_Months'],bins=bins,labels=label)
```

```python
df
```

```python
df['Senior_flag']=np.where(df['Age']>=60,'Senior','Adult')
```

```python
df['Churn_flag']=df['Churn'].map({'Yes':1,'No':0})
```

```python
df.info()
```

```python
df['Last_Interaction_Date'].fillna(df['Last_Interaction_Date'].median(),inplace=True)
```

```python
df.to_csv('Clean_churn_data.csv',index=False)
```
